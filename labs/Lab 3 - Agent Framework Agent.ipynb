# Lab 3 — Agent Framework Agent

Build a personal **AI travel agent** with the **Microsoft Agent Framework**. The agent dreams up a random vacation spot and instantly creates a detailed day-trip itinerary for it. Along the way you will:

1. Configure a chat client.
2. Give the agent a **custom tool** (a plain Python function).
3. Create an `Agent`, run it, and read the result.

> This lab runs on your **Microsoft Foundry** project — the same endpoint and model you set up in **Lab 1**. Make sure you completed **Lab 1** first.

## Prerequisites: configure your environment

This lab reuses the same `.env` values from **Lab 1** — no new configuration needed:

```
FOUNDRY_PROJECT_ENDPOINT=https://<resource>.services.ai.azure.com/api/projects/<project>
AZURE_AI_MODEL_DEPLOYMENT_NAME=gpt-4.1
```

- `FOUNDRY_PROJECT_ENDPOINT` — your Foundry project endpoint.
- `AZURE_AI_MODEL_DEPLOYMENT_NAME` — your deployed chat model.

Make sure you are signed in with `az login` so `DefaultAzureCredential` can authenticate.

## 1. Set up the environment

Import the required libraries. `agent_framework` provides the core building blocks, while `dotenv` helps us manage configuration securely.

````python
import os
from random import randint
from dotenv import load_dotenv

# Import the core agent components
from agent_framework import Agent
from agent_framework.foundry import FoundryChatClient
from azure.identity import DefaultAzureCredential

# Load environment variables from the .env file
load_dotenv()
````

## 2. Create a custom tool

Define a function the agent can call as a **tool**. In Python, the function's docstring (`"""..."""`) tells the agent what the function is for.

````python
def get_random_destination() -> str:
    """Get a random vacation destination."""
    destinations = [
        "Barcelona, Spain", "Paris, France", "Berlin, Germany",
        "Tokyo, Japan", "Sydney, Australia", "New York, USA",
        "Cairo, Egypt", "Cape Town, South Africa",
        "Rio de Janeiro, Brazil", "Bali, Indonesia"
    ]
    return destinations[randint(0, len(destinations) - 1)]
````

## 3. Build and run the agent

Initialize the `FoundryChatClient` from the environment variables, then create an `Agent` with instructions and the `get_random_destination` tool. Finally, give it a prompt and let it run.

````python
# Initialize the chat client pointing to your Foundry project
chat_client = FoundryChatClient(
    project_endpoint=os.environ["FOUNDRY_PROJECT_ENDPOINT"],
    model=os.environ["AZURE_AI_MODEL_DEPLOYMENT_NAME"],
    credential=DefaultAzureCredential(),
)

# Create the agent
agent = Agent(
    client=chat_client,
    instructions="You are a helpful AI Agent that can help plan vacations for customers at random destinations.",
    tools=[get_random_destination],
)

# Run the agent with a prompt
response = await agent.run("Plan me a day trip")
````

The `response` object contains the full conversation history. To display the final travel plan, extract the text from the agent's last message.

````python
# Get the text from the last message in the response
last_message = response.messages[-1]
text_content = last_message.contents[0].text
print("Travel plan:")
print(text_content)
````

You just built a functional AI agent: you gave it a purpose, equipped it with a custom tool to perform an action, and set it to work on a task. This is the foundation for more complex agents.
