## 2. Configure your project

Copy `.env.example` to `.env` and fill in your values, then run the cell below.

```
FOUNDRY_PROJECT_ENDPOINT=https://<resource>.services.ai.azure.com/api/projects/<project>
AZURE_AI_MODEL_DEPLOYMENT_NAME=gpt-4.1
```

````python
import os
from dotenv import load_dotenv

load_dotenv()

project_endpoint = os.environ["FOUNDRY_PROJECT_ENDPOINT"]
model = os.environ.get("AZURE_AI_MODEL_DEPLOYMENT_NAME", "gpt-4.1")

print("Project endpoint:", project_endpoint)
print("Model deployment:", model)
````

## 3. Authenticate and test

`AIProjectClient` uses `DefaultAzureCredential`, which picks up your `az login` session. We get an OpenAI-compatible client from the project and run a single chat completion to confirm everything works.

````python
from azure.ai.projects import AIProjectClient
from azure.identity import DefaultAzureCredential

project = AIProjectClient(endpoint=project_endpoint, credential=DefaultAzureCredential())
openai_client = project.get_openai_client()

completion = openai_client.chat.completions.create(
    model=model,
    messages=[{"role": "user", "content": "Reply with exactly: connection OK"}],
)
print(completion.choices[0].message.content)
````

If you see `connection OK`, your environment is ready. Continue to **Lab 2 — Prompt Agent**.
